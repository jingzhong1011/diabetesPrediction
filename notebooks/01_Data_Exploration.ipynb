# 01. Exploratory Data Analysis (EDA)

This notebook focuses on understanding the distribution of the clinical features within the diabetes dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append('../')
from src.data_processing import load_and_clean_data

plt.style.use('seaborn-v0_8')
sns.set_palette("pastel")
%matplotlib inline
```

## Load Dataset
*(Here we generate dummy data representing the clinical dataset for demonstration purposes. In production, uncomment the load function to read `data_pima.csv`.)*

```python
# df = load_and_clean_data('../data/raw/data_pima.csv', fill_method='mean')

np.random.seed(42)
df = pd.DataFrame({
    'Glucose': np.random.normal(120, 30, 500),
    'BloodPressure': np.random.normal(70, 10, 500),
    'Insulin': np.random.exponential(50, 500),
    'Age': np.random.randint(21, 80, 500),
    'Outcome': np.random.randint(0, 2, 500)
})
df.head()
```

## Feature Distributions
We visualize the density and spread of key health indicators.

```python
features_to_plot = ['Glucose', 'BloodPressure', 'Insulin', 'Age']
colors = ['lightcoral', 'lightsalmon', 'darkseagreen', 'lightsteelblue']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for i, (col, color) in enumerate(zip(features_to_plot, colors)):
    sns.histplot(df[col], kde=True, color=color, ax=axes[i])
    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Density')

plt.tight_layout()
plt.show()
```
